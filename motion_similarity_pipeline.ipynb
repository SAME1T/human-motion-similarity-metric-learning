# 1. Test Videolarından MediaPipe ile 2D Koordinat Çıkarımı

Bu adımda test videolarından 17 adet bel üstü noktanın X ve Y koordinatları çıkarılacaktır. 
MediaPipe kullanılarak her video frame bazlı analiz edilir. Çıktılar `.npy` formatında `npy_test_cikti` klasörüne yazılır.


```python
# Gerekli kütüphaneler
import os
import cv2
import numpy as np
import mediapipe as mp

# Ana proje klasörü
BASE = r"C:\Users\The Coder Farmer\Desktop\Autoencoder_video"

# Test video klasörü ve çıktı dizini
DIR_TEST_VIDEO = os.path.join(BASE, "test_video")
DIR_TEST_OUT = os.path.join(BASE, "npy_test_cikti")
os.makedirs(DIR_TEST_OUT, exist_ok=True)

# MediaPipe hazırla
mp_pose = mp.solutions.pose
pose = mp_pose.Pose(static_image_mode=False, model_complexity=1)

# Bel üstü 17 nokta (33 noktadan seçilenler)
BEL_USTU_IDX = [
    0, 11, 12, 13, 14, 15, 16, 23, 24  # Baş, omuzlar, dirsekler, bilekler, kalça
]
```

```python
# Test videoları listesi (dosya adları "61.mp4" gibi)
test_videolar = ["61.mp4", "62.mp4", "63.mp4", "64.mp4"]

# Her video için koordinat çıkarımı
for name in test_videolar:
    path = os.path.join(DIR_TEST_VIDEO, name)
    cap = cv2.VideoCapture(path)

    koord_listesi = []

    while cap.isOpened():
        ok, frame = cap.read()
        if not ok:
            break

        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = pose.process(frame_rgb)

        if results.pose_landmarks:
            keypoints = results.pose_landmarks.landmark
            frame_data = []

            for idx in BEL_USTU_IDX:
                frame_data.extend([keypoints[idx].x, keypoints[idx].y])

            koord_listesi.append(frame_data)

    cap.release()

    # numpy array'e çevir ve .npy olarak kaydet
    dizi = np.array(koord_listesi)
    out_path = os.path.join(DIR_TEST_OUT, name.replace(".mp4", ".npy"))
    np.save(out_path, dizi)
    print(f"{name} → {dizi.shape} boyutlu npy çıkarıldı.")
```

# 3 Eğitim ve Test Videolarından Sağ Kol Açılarının Çıkarımı

Her frame için 3 açı hesaplanacaktır:  
- Dirsek Açısı (omuz–dirsek–bilek)  
- Gövde–Kol Açısı (kalça–omuz–dirsek)  
- Omuz Hattı Açısı (sol omuz–sağ omuz–sağ dirsek)  

Elde edilen açılar `.npy` formatında `npy_aci_cikti/` ve `npy_aci_test_cikti/` klasörlerine ayrı ayrı kaydedilecektir.


```python
#  3 nokta ile açı hesaplama fonksiyonu(yardımcı fonksiyon)
import math

def calculate_angle(p1, p2, p3):
    """
    Üç nokta arasındaki açıyı hesaplar (derece cinsinden).
    Noktalar (x, y) tuple'larıdır. Açılar 0–180 derece arasıdır.
    """
    a = np.array(p1)
    b = np.array(p2)
    c = np.array(p3)

    ba = a - b
    bc = c - b

    cosine_angle = np.dot(ba, bc) / (np.linalg.norm(ba) * np.linalg.norm(bc) + 1e-8)
    angle = np.arccos(np.clip(cosine_angle, -1.0, 1.0))
    return np.degrees(angle)
```

```python
DIR_ANG_TRAIN = os.path.join(BASE, "npy_aci_cikti")
os.makedirs(DIR_ANG_TRAIN, exist_ok=True)
# eğitim açı klasörü oluşturma
```

```python
# eğitim videoları için açı çıkarımı
for name in egitim_videolar:
    npy_path = os.path.join(DIR_EGITIM_OUT, name.replace(".mp4", ".npy"))
    keypoints = np.load(npy_path)

    acilar = []

    for frame in keypoints:
        # Noktaları sırayla çekiyoruz
        # 17 nokta x,y → 34 elemanlı vektör
        points = np.reshape(frame, (-1, 2))

        # Gerekli indeksler
        RH = points[8]   # Sağ kalça
        RS = points[1]   # Sağ omuz
        RE = points[3]   # Sağ dirsek
        RW = points[5]   # Sağ bilek
        LS = points[2]   # Sol omuz

        # 3 açıyı hesapla
        a_dirsek = calculate_angle(RS, RE, RW)
        a_govde = calculate_angle(RH, RS, RE)
        a_omuzhat = calculate_angle(LS, RS, RE)

        acilar.append([a_dirsek, a_govde, a_omuzhat])

    acilar = np.array(acilar)
    np.save(os.path.join(DIR_ANG_TRAIN, name.replace(".mp4", "_ang.npy")), acilar)
    print(f"{name} için açı verisi çıkarıldı → {acilar.shape}")
```

# 5. PENCERELEME (WINDOWING)

Koordinat (34 boyut) ve açı (3 boyut) verileri birleştirilerek, her video sliding window yöntemiyle sabit uzunlukta pencerelere bölünür.  
Bu pencereler modelin eğitimine ve testine veri sağlayacak biçimde kaydedilir.  

Çıktılar `motion_similarity_pipeline/` klasörüne `.npy` dosyası olarak kaydedilir.


```python
import os
import numpy as np

# === TEMEL PARAMETRELER ===
WINDOW_SIZE = 30  # Pencere uzunluğu
FEATURE_DIM = 34 + 3  # 17 nokta (X,Y) + 3 açı = 37 özellik

# === ANA DİZİN ===
BASE = r"C:\Users\The Coder Farmer\Desktop\Autoencoder_video"
WINDOW_OUT_DIR = os.path.join(BASE, "motion_similarity_pipeline")  # ✅ Pencere çıktıları buraya
os.makedirs(WINDOW_OUT_DIR, exist_ok=True)

# === GİRDİ KLASÖRLERİ ===
DIR_EGITIM_NPY = os.path.join(BASE, "npy_cikti")
DIR_TEST_NPY   = os.path.join(BASE, "npy_test_cikti")
DIR_EGITIM_ANG = os.path.join(BASE, "npy_aci_cikti")
DIR_TEST_ANG   = os.path.join(BASE, "npy_aci_test_cikti")

# === KAYIT YOLLARI ===
PATH_OUT_TRAIN = os.path.join(WINDOW_OUT_DIR, f"train_windows_{WINDOW_SIZE}.npy")
PATH_OUT_TEST  = os.path.join(WINDOW_OUT_DIR, f"test_windows_{WINDOW_SIZE}.npy")
```

```python
# pencereleme fonksiyonu
def sliding_window(data, window_size):
    """
    Frame dizisini sabit uzunlukta pencerelere böler.

    Args:
        data (ndarray): (frame_sayısı, özellik_sayısı)
        window_size (int): pencere uzunluğu

    Returns:
        ndarray: (pencere_sayısı, window_size, özellik_sayısı)
    """
    windows = []
    for i in range(0, len(data) - window_size + 1):
        window = data[i:i + window_size]
        windows.append(window)
    return np.array(windows)
```

```python
# eğitim pencereleri oluşturma
egitim_windows = []

for i in range(1, 61):  # 1.mp4 → 60.mp4
    coord_path = os.path.join(DIR_EGITIM_NPY, f"{i}.npy")
    angle_path = os.path.join(DIR_EGITIM_ANG, f"{i}_ang.npy")

    coords = np.load(coord_path)   # (F, 34)
    angles = np.load(angle_path)   # (F, 3)

    # Koordinat ve açıları birleştir
    birlesik = np.concatenate([coords, angles], axis=1)  # (F, 37)

    # Pencerelere böl
    pencere = sliding_window(birlesik, WINDOW_SIZE)
    egitim_windows.append(pencere)

# Tüm eğitim pencerelerini birleştir
egitim_windows = np.concatenate(egitim_windows, axis=0)

# .npy dosyasına kaydet
np.save(PATH_OUT_TRAIN, egitim_windows)
print(f"[✓] Eğitim pencereleri kaydedildi → {PATH_OUT_TRAIN}")
print(f"    Boyut: {egitim_windows.shape}")
```

```python
# test pencereleri oluşturma
test_windows = []

for i in range(61, 65):  # 61.mp4 → 64.mp4
    coord_path = os.path.join(DIR_TEST_NPY, f"{i}.npy")
    angle_path = os.path.join(DIR_TEST_ANG, f"{i}_ang.npy")

    coords = np.load(coord_path)
    angles = np.load(angle_path)

    birlesik = np.concatenate([coords, angles], axis=1)
    pencere = sliding_window(birlesik, WINDOW_SIZE)
    test_windows.append(pencere)

test_windows = np.concatenate(test_windows, axis=0)
np.save(PATH_OUT_TEST, test_windows)
print(f"[✓] Test pencereleri kaydedildi → {PATH_OUT_TEST}")
print(f"    Boyut: {test_windows.shape}")
```

# 6. Temel Autoencoder Eğitimi

Bu aşamada, 30 framelik pencerelerden oluşan giriş dizisi ile basit bir autoencoder modeli eğitilecektir.  
Girdi: (30 frame, 37 özellik)  
Latent boyutu: 32  
Kayıp fonksiyonu: MSE (yeniden yapılandırma hatası)

Eğitim tamamlandıktan sonra tüm eğitim verisinin latent gösterimleri `latent_train_base/` klasörüne `.npy` dosyası olarak kaydedilecektir.


```python
import os
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

# ========== DOSYA YOLLARI ==========
BASE = r"C:\Users\The Coder Farmer\Desktop\Autoencoder_video"
PIPE_DIR = os.path.join(BASE, "motion_similarity_pipeline")

# Eğitim verisi .npy dosyası
PATH_TRAIN = os.path.join(PIPE_DIR, "train_windows_30.npy")

# Latent çıktı klasörü
LATENT_OUT_DIR = os.path.join(PIPE_DIR, "latent_train_base")
os.makedirs(LATENT_OUT_DIR, exist_ok=True)

# ========== MODEL SINIFI ==========
class AutoEncoder(nn.Module):
    def __init__(self, input_dim, latent_dim=32):
        super(AutoEncoder, self).__init__()
        
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),
            nn.Linear(512, latent_dim)
        )
        
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 512),
            nn.ReLU(),
            nn.Linear(512, input_dim)
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return decoded, encoded
```

```python
# ========== VERİYİ YÜKLE ==========
X_train = np.load(PATH_TRAIN)  # (N, 30, 37)
X_train = X_train.reshape(X_train.shape[0], -1)  # → (N, 1110) gibi

# Giriş boyutunu belirle
input_dim = X_train.shape[1]
latent_dim = 32

# Torch tensöre çevir
X_tensor = torch.tensor(X_train, dtype=torch.float32)

# DataLoader
batch_size = 128
train_dataset = TensorDataset(X_tensor)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
```

```python
# ========== MODEL, OPTİMİZER, LOSS ==========
model = AutoEncoder(input_dim=input_dim, latent_dim=latent_dim)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.MSELoss()

# ========== EĞİTİM ==========
epochs = 70
for epoch in range(epochs):
    model.train()
    total_loss = 0.0

    for batch in train_loader:
        x = batch[0]
        optimizer.zero_grad()
        out, _ = model(x)
        loss = criterion(out, x)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * x.size(0)

    avg_loss = total_loss / len(train_loader.dataset)
    print(f"Epoch {epoch+1:02d}/{epochs} - Loss: {avg_loss:.6f}")
```

```python
# cosine ve dtw benzerlik ölçümü
def cosine_sim(a, b):
    return 1 - cosine(a, b)

def mean_cosine(v1, v2):
    min_len = min(len(v1), len(v2))
    return np.mean([cosine_sim(v1[i], v2[i]) for i in range(min_len)])

# Her test videosu için 60 eğitim videosuna karşı benzerlik ölç
for test_id in range(61, 65):
    test_latent = np.load(os.path.join(LATENT_OUT_TEST, f"latent_{test_id}.npy"))
    print(f"\nTest Video {test_id}.mp4 için en benzer eğitim videoları:")

    similarities = []
    for train_id in range(1, 61):
        train_latent = np.load(os.path.join(LATENT_OUT_TRAIN, f"latent_{train_id}.npy"))

        # Cosine benzerlik
        cos = mean_cosine(test_latent, train_latent)

        # DTW benzerlik
        dist, _ = fastdtw(test_latent, train_latent)

        similarities.append((train_id, cos, dist))

    # En iyi eşleşmeleri sırala
    similarities.sort(key=lambda x: -x[1])  # Cosine’a göre büyükten küçüğe
    print("  [Cosine] En benzer ilk 3:", similarities[:3])

    similarities.sort(key=lambda x: x[2])  # DTW’ye göre küçükten büyüğe
    print("  [DTW]    En benzer ilk 3:", similarities[:3])
```

# 9. MNA Autoencoder Eğitimi

Bu modelde giriş verisi iki ayrı kola ayrılır:
- Koordinat vektörü (34 boyut): 17 eklem noktası (X, Y)
- Açı vektörü (3 boyut): Dirsek, gövde–kol ve omuz hizası

Her biri kendi encoder bloğundan geçerek tek bir latent vektör haline getirilir.
Bu birleşik temsil decoder tarafından tekrar yapılandırılır.
Amaç: Daha ayrıştırıcı (disentangled) bir temsil öğrenmektir.


```python
# Yeni latent çıkış klasörü
LATENT_OUT_MNA = os.path.join(PIPE_DIR, "latent_train_mna")
os.makedirs(LATENT_OUT_MNA, exist_ok=True)

# Eğitim verisi (birleştirilmiş)
X_train = np.load(os.path.join(PIPE_DIR, "train_windows_30.npy"))  # (N, 30, 37)
```

```python
# veriyi kordinat ve açı olarak ayır

# 34: koordinat (17*2), 3: açı → ayrıştır
X_coord = X_train[:, :, :34]  # (N, 30, 34)
X_angle = X_train[:, :, 34:]  # (N, 30, 3)

# Flatten (batch, window, feature) → (batch, window*feature)
X_coord = X_coord.reshape(X_coord.shape[0], -1)  # (N, 1020)
X_angle = X_angle.reshape(X_angle.shape[0], -1)  # (N, 90)

# Torch tensöre çevir
X_coord = torch.tensor(X_coord, dtype=torch.float32)
X_angle = torch.tensor(X_angle, dtype=torch.float32)

# Dataset & Loader
train_dataset = TensorDataset(X_coord, X_angle)
train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True)
```

```python
# mna model tanımı
class MNA_AutoEncoder(nn.Module):
    def __init__(self, coord_input=1020, angle_input=90, latent_dim=32):
        super(MNA_AutoEncoder, self).__init__()

        # Koordinat encoder
        self.coord_encoder = nn.Sequential(
            nn.Linear(coord_input, 256),
            nn.ReLU(),
            nn.Linear(256, latent_dim // 2)
        )

        # Açı encoder
        self.angle_encoder = nn.Sequential(
            nn.Linear(angle_input, 128),
            nn.ReLU(),
            nn.Linear(128, latent_dim // 2)
        )

        # Decoder (birleşik latent vektör kullanır)
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 512),
            nn.ReLU(),
            nn.Linear(512, coord_input + angle_input)
        )

    def forward(self, x_coord, x_angle):
        z_coord = self.coord_encoder(x_coord)   # (N, latent/2)
        z_angle = self.angle_encoder(x_angle)   # (N, latent/2)
        z = torch.cat([z_coord, z_angle], dim=1)  # (N, latent)

        reconstructed = self.decoder(z)
        return reconstructed, z
```

```python
# Giriş boyutlarını otomatik al
coord_input = X_coord.shape[1]
angle_input = X_angle.shape[1]
latent_dim = 32

# Model
model = MNA_AutoEncoder(coord_input=coord_input, angle_input=angle_input, latent_dim=latent_dim)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

# Eğitim
epochs = 70
for epoch in range(epochs):
    model.train()
    total_loss = 0.0

    for x_c, x_a in train_loader:
        optimizer.zero_grad()
        output, _ = model(x_c, x_a)

        # Girişin kendisi hedef
        target = torch.cat([x_c, x_a], dim=1)
        loss = criterion(output, target)

        loss.backward()
        optimizer.step()
        total_loss += loss.item() * x_c.size(0)

    avg_loss = total_loss / len(train_loader.dataset)
    print(f"Epoch {epoch+1}/{epochs} - Loss: {avg_loss:.6f}")
```

# 13. MNA Benzerlik Kıyaslaması

Test videolarının latent temsilleri, eğitim latentleri ile karşılaştırılır.  
Benzerlik hesaplamasında hem Cosine Similarity hem DTW kullanılacaktır.  
Amaç: MNA modelinin klasik AE’ye kıyasla daha iyi ayrım yapıp yapmadığını test etmektir.


```python
# tüm süreci içeren benzerlik döngüsü
def cosine_sim(a, b):
    from scipy.spatial.distance import cosine
    return 1 - cosine(a, b)

def mean_cosine(v1, v2):
    min_len = min(len(v1), len(v2))
    return np.mean([cosine_sim(v1[i], v2[i]) for i in range(min_len)])

from fastdtw import fastdtw

print("=== MNA Model Benzerlik Skorları ===")

for test_id in range(61, 65):
    test_latent_path = os.path.join(PIPE_DIR, "latent_test_mna", f"latent_{test_id}.npy")
    test_latent = np.load(test_latent_path)

    similarities = []
    for train_id in range(1, 61):
        train_latent_path = os.path.join(PIPE_DIR, "latent_train_mna", f"latent_{train_id}.npy")
        train_latent = np.load(train_latent_path)

        # Cosine
        cosine_score = mean_cosine(test_latent, train_latent)

        # DTW
        dtw_score, _ = fastdtw(test_latent, train_latent)

        similarities.append((train_id, cosine_score, dtw_score))

    # Cosine sıralama
    similarities.sort(key=lambda x: -x[1])
    top_cosine = similarities[:3]

    # DTW sıralama
    similarities.sort(key=lambda x: x[2])
    top_dtw = similarities[:3]

    print(f"\nTest Video {test_id}.mp4")
    print("  [Cosine] En benzer 3:", top_cosine)
    print("  [DTW]    En benzer 3:", top_dtw)
```

# 14 . KNN ile başarı yüzdesi (AE ve MNA karşılaştırmalı)
Latent vektörler, ait oldukları video ID’si (1..60) ile etiketlenir.
- Eğitim pencereleri, her video için train/val olarak ayrılır (ör. %80 / %20).
- Train set ile KNN modeli kurulur; val set üzerinde doğruluk (accuracy %) ölçülür.
- Ayrıca Recall@K (K=1 ve K=5) hesaplanır.
- Sonuçlar `motion_similarity_pipeline/metrics/knn_*.csv` olarak kaydedilir.


```python
import os, numpy as np, pandas as pd, random
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# ==== YOLLAR ====
BASE = r"C:\Users\The Coder Farmer\Desktop\Autoencoder_video"
PIPE_DIR = os.path.join(BASE, "motion_similarity_pipeline")
METRICS_DIR = os.path.join(PIPE_DIR, "metrics")
os.makedirs(METRICS_DIR, exist_ok=True)

LATENT_DIR_AE  = os.path.join(PIPE_DIR, "latent_train_base")  # AE latent_1.npy ... latent_60.npy
LATENT_DIR_MNA = os.path.join(PIPE_DIR, "latent_train_mna")   # MNA latent_1.npy ... latent_60.npy

# ==== KNN PARAMETRELERİ ====
K_NEIGHBORS = 5          # değiştirilebilir
METRIC = "cosine"        # "euclidean" veya "cosine" önerilir
VAL_RATIO = 0.20         # her video için %20 validation
STANDARDIZE = True       # latent'leri ölçeklemek genelde faydalı
SEED = 42                # tekrarlanabilirlik

random.seed(SEED)
np.random.seed(SEED)

def load_latents(latent_dir, n_videos=60):
    """
    latent_dir içinde latent_{i}.npy dosyalarını yükler.
    Her dosya: (pencere, latent_dim)
    Dönen:
      X: (N, D) latent vektörleri
      y: (N,) video id (1..n_videos)
      idx_slices: her video için kaç pencere olduğu (list)
    """
    X_list, y_list, counts = [], [], []
    for vid in range(1, n_videos+1):
        f = os.path.join(latent_dir, f"latent_{vid}.npy")
        if not os.path.exists(f):
            counts.append(0)
            continue
        arr = np.load(f)
        if arr.ndim == 1:
            arr = arr[None, :]
        X_list.append(arr)
        y_list.append(np.full(len(arr), vid, dtype=int))
        counts.append(len(arr))
    if not X_list:
        raise RuntimeError(f"Latent bulunamadı: {latent_dir}")
    X = np.vstack(X_list)
    y = np.concatenate(y_list)
    return X, y, counts

def split_train_val_per_video(X, y, counts, val_ratio=0.2):
    """
    Her video sınıfı içinde rasgele train/val ayırır.
    Döner: Xtr, ytr, Xval, yval
    """
    Xtr, ytr, Xval, yval = [], [], [], []
    idx = 0
    for vid, c in enumerate(counts, start=1):
        if c == 0:
            continue
        inds = np.arange(idx, idx+c)
        idx += c
        np.random.shuffle(inds)
        n_val = max(1, int(round(val_ratio * c)))
        val_inds = inds[:n_val]
        tr_inds  = inds[n_val:]
        Xval.append(X[val_inds]); yval.append(y[val_inds])
        Xtr.append(X[tr_inds]);   ytr.append(y[tr_inds])
    Xtr = np.vstack([a for a in Xtr if a.size])
    ytr = np.concatenate([a for a in ytr if a.size])
    Xval = np.vstack([a for a in Xval if a.size])
    yval = np.concatenate([a for a in yval if a.size])
    return Xtr, ytr, Xval, yval

def recall_at_k(train_X, train_y, val_X, val_y, k=5, metric="cosine"):
    """
    Basit Recall@K: val örneğinin en yakın k komşusu içinde doğru sınıf var mı?
    KNN skorundan bağımsız, komşuluk listesi ile hesaplanır.
    """
    # fit KNN yalnızca komşuları almak için
    knn = KNeighborsClassifier(n_neighbors=k, metric=metric)
    knn.fit(train_X, train_y)
    # komşu indeksleri
    neigh_inds = knn.kneighbors(val_X, return_distance=False)  # (Nval, k)
    neigh_labels = train_y[neigh_inds]                         # (Nval, k)
    hits = (neigh_labels == val_y[:, None]).any(axis=1)
    return round(100.0 * hits.mean(), 2)
```

```python
def evaluate_knn_for_model(model_name, latent_dir, k=5, metric="cosine",
                           val_ratio=0.20, standardize=True):
    # Latentleri yükle
    X, y, counts = load_latents(latent_dir, n_videos=60)

    # Train/Val böl
    Xtr, ytr, Xval, yval = split_train_val_per_video(X, y, counts, val_ratio=val_ratio)

    # Ölçekle (sadece train ile fit, val'e uygula)
    scaler = None
    if standardize:
        scaler = StandardScaler()
        Xtr = scaler.fit_transform(Xtr)
        Xval = scaler.transform(Xval)

    # KNN eğit
    knn = KNeighborsClassifier(n_neighbors=k, metric=metric)
    knn.fit(Xtr, ytr)

    # Doğruluk
    ypred = knn.predict(Xval)
    acc = round(100.0 * accuracy_score(yval, ypred), 2)

    # Recall@1 ve Recall@5
    r1 = recall_at_k(Xtr, ytr, Xval, yval, k=1, metric=metric)
    r5 = recall_at_k(Xtr, ytr, Xval, yval, k=5, metric=metric)

    # Per-video doğruluk raporu
    # val setini video bazında ayırıp her video için accuracy hesaplayalım
    # yval içinde sınıf etiketleri var → bunları gruplayıp hesaplarız
    df = pd.DataFrame({"y_true": yval, "y_pred": ypred})
    per_video = (df.groupby("y_true")
                   .apply(lambda d: round(100.0 * (d["y_true"] == d["y_pred"]).mean(), 2))
                   .reset_index())
    per_video.columns = ["video_id", "acc_percent"]

    # Kayıt
    out_csv = os.path.join(METRICS_DIR, f"knn_{model_name.lower()}_per_video.csv")
    per_video.to_csv(out_csv, index=False)

    summary_row = pd.DataFrame([{
        "model": model_name,
        "k": k,
        "metric": metric,
        "val_ratio": val_ratio,
        "standardize": standardize,
        "accuracy_percent": acc,
        "recall@1_percent": r1,
        "recall@5_percent": r5,
        "n_train": len(ytr),
        "n_val": len(yval)
    }])
    summary_path = os.path.join(METRICS_DIR, "knn_summary.csv")
    if os.path.exists(summary_path):
        prev = pd.read_csv(summary_path)
        pd.concat([prev, summary_row], ignore_index=True).to_csv(summary_path, index=False)
    else:
        summary_row.to_csv(summary_path, index=False)

    print(f"[{model_name}] KNN Accuracy: %{acc} | R@1: %{r1} | R@5: %{r5}")
    print(f"Per-video CSV → {out_csv}")
    return acc, r1, r5
```

```python
ae_acc, ae_r1, ae_r5   = evaluate_knn_for_model("AE",  LATENT_DIR_AE,
                                                 k=K_NEIGHBORS, metric=METRIC,
                                                 val_ratio=VAL_RATIO, standardize=STANDARDIZE)

mna_acc, mna_r1, mna_r5 = evaluate_knn_for_model("MNA", LATENT_DIR_MNA,
                                                 k=K_NEIGHBORS, metric=METRIC,
                                                 val_ratio=VAL_RATIO, standardize=STANDARDIZE)

print("\n=== ÖZET ===")
print(f"AE : Acc %{ae_acc} | R@1 %{ae_r1} | R@5 %{ae_r5}")
print(f"MNA: Acc %{mna_acc} | R@1 %{mna_r1} | R@5 %{mna_r5}")
print(f"Toplam özet: {os.path.join(METRICS_DIR, 'knn_summary.csv')}")
```
